```python
import os
from dotenv import load_dotenv, find_dotenv
import requests
import pandas as pd
import matplotlib.pyplot as plt

assert load_dotenv(find_dotenv("fredapi.env")), "no .env found"

key = os.environ['FRED_API_KEY']

url = "https://api.stlouisfed.org/fred/series/observations"
params = {'series_id': 'IORB' , 'api_key': key, 'file_type': 'json'}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    df = pd.DataFrame(data)

else:
    print(f'Error: {response.status_code}')
```

```python
df
```

```python
df['realtime_start'].unique(), df['realtime_end'].unique(), df['observation_start'].unique(), df['observation_end'].unique(), df['units'].unique(), df['output_type'].unique(), df['file_type'].unique(), df['order_by'].unique(), df['sort_order'].unique(), df['count'].unique(), df['offset'].unique(), df['limit'].unique()
```

```python
df = df.drop(columns=['realtime_start', 'realtime_end', 'observation_start', 'observation_end', 'units', 'output_type', 'file_type', 'order_by', 'sort_order', 'count', 'offset', 'limit'])
```

```python
df = pd.json_normalize(df['observations'])
```

```python
df = df.drop(columns=['realtime_start', 'realtime_end'])
```

```python
df['date'] = pd.to_datetime(df['date'])
df['value'] = pd.to_numeric(df['value'], errors = 'coerce')
```

```python
df.set_index('date')
```

```python
plt.plot(df['date'], df['value'])
```

```python
df1 = pd.json_normalize(df1['observations'])
```

```python
iorrparams = {'series_id': 'IORR' , 'api_key': key, 'file_type': 'json'}

response = requests.get(url, params=iorrparams)

if response.status_code == 200:
    data = response.json()

    df2 = pd.DataFrame(data)

else:
    print(f'Error: {response.status_code}')

df2 = df2.drop(columns=['realtime_start', 'realtime_end', 'observation_start', 'observation_end', 'units', 'output_type', 'file_type', 'order_by', 'sort_order', 'count', 'offset', 'limit'])
df2 = pd.json_normalize(df2['observations'])

df2 = df2.drop(columns=['realtime_start', 'realtime_end'])
df2['date'] = pd.to_datetime(df2['date'])
df2['value'] = pd.to_numeric(df2['value'], errors = 'coerce')
df2.set_index('date')
```
