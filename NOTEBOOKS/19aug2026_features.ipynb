```python
import json
import pandas as pd
from reserve_ampleness.data.registry import resolve
from reserve_ampleness.data.nyfed_load_rates import load_rates
```

```python
df = load_rates("nyfed_all_rates", ["SOFR"])
```

```python
calendar = pd.DataFrame(index=df.index)

calendar["day"] = calendar.index.day_name()
```

```python
s = calendar.index.to_series()

calendar["turn_nights"] = (s.shift(-1) - s).dt.days
```

```python
calendar
```

```python
month = calendar.index.month


calendar["month"] = month
```

```python
calendar['is_month_end'] = (calendar["month"].shift(-1) != calendar["month"])
```

```python
calendar['is_quarter_end'] = (calendar['month'].isin([3, 6, 9, 12])) & (calendar['is_month_end'])
```

```python
calendar['is_year_end'] = (calendar['month'] == 12) & (calendar['is_quarter_end'])
```
