```python
import pandas as pd
import requests

url = "https://markets.newyorkfed.org/api/rates/secured/sofr/search.json"
params = {'startDate': '2000-01-01' , 'endDate': '2026-08-10'}

response = requests.get(url, params=params)

if response.status_code == 200:
    data = response.json()

    df = pd.DataFrame(data)

else:
    print(f'Error: {response.status_code}')
```

```python
df.head(10)
```

```python
flat = pd.json_normalize(df['refRates'])
```

```python
flat.head(10)
```

```python
flat['effectiveDate'] = pd.to_datetime(flat['effectiveDate'])
```

```python
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

plt.figure(figsize=(15, 4))

plt.plot(flat['effectiveDate'], flat['percentRate'])
```

```python
plt.figure(figsize=(15, 4))
p99 = flat['percentPercentile99']
p1 = flat['percentPercentile1']
spread = p99 - p1

plt.plot(flat['effectiveDate'], spread)
```
